# Writing reports

Project documentation and writing reports in a clear and concise manner is an important skill for any engineer or scientist. The reader should always be in mind. The goal is to communicate your findings effectively and efficiently. We shall here go through some tools and best practices for writing reports.

## Why not just use Word or Google Docs?

While Word and Google Docs are popular tools for writing documents, they have several limitations when it comes to technical writing, especially in fields like engineering and science. Mathematical notation and code snippets are poorly supported. Referecing figures, tables, and equations feels like manual labor. Lack of good version control, ever used "Track Changes" in Word? It is a nightmare!

Instead, we recommend using tools like LaTeX with Quarto, which are designed for technical writing. They offer superior support for mathematical notation, code integration, and document structuring. Additionally, they work well with version control systems like Git, making collaboration and tracking changes much easier.

## Why not use Overleaf?

You can use Overleaf, an online LaTeX editor, for writing reports. However, it has some limitations compared to using LaTeX or Quarto locally. Overleaf requires a paid subscription for any serious work. It also has limited customization options and may not support all LaTeX packages. Additionally, working locally allows for better integration with version control systems like Git.

## Recommended tools

This guide will focus on using Quarto and Jupyter Notebooks for writing reports. Quarto is a powerful tool that allows you to create dynamic documents that can include code, text, and visualizations. Jupyter Notebooks are great for combining code and narrative text in a single document.

We will write using [markdown](https://quarto.org/docs/authoring/markdown-basics.html), which is a lightweight markup language that is easy to learn and use. Quarto extends markdown with additional features for technical writing. 

Markdown and Quarto support $\LaTeX$ for mathematical notation. In this guide, you will learn the basics of $\LaTeX$ and how to use it in the context of writing reports with Quarto and markdown. See @sec-what_is_latex for a brief introduction to $\LaTeX$.

The tools we will use are: VS Code, Quarto, Jupyter Notebooks and python for coding and generating plots.

### The workflow

In VS-Code you create a Jupyter Notebook (.ipynb). You write your code and text in the notebook or markdown file. When you are ready to generate the report, you "render" the notebook or markdown file to produce a formatted document (usually PDF or HTML).

````bash
quarto render my_report.ipynb --to pdf
````

## What is $\LaTeX$? {#sec-what_is_latex}

$\LaTeX$ is a typesetting system that is widely used for technical and scientific documents. It allows you to create documents with complex mathematical notation and formatting. $\LaTeX$ uses a markup language to define the structure and content of the document, which is then compiled into a formatted output (usually PDF).

Typically LateX documents have the file extension `.tex` and are edited using specialized editors like TeXShop, TeXworks, or Overleaf. The former mentioned require large [installation and setup](https://miktex.org/download). Regardless, all these solutions require to write a ton of [boilerplate code](https://en.wikipedia.org/wiki/Boilerplate_code) to generate a [simple document](https://samrobbins.uk/blog/my-latex-boilerplate/). However, in Quarto and markdown, you can include $\LaTeX$ code directly within your markdown files or Jupyter Notebooks with little to no bioler plate code.

### YAML Front Matter

The front matter controls document metadata and formatting options:

````yaml
---
title: "Numerical Integration: Trapezoidal Rule vs Gaussian Quadrature"
author: "Student Name"
date: today
format:
  pdf:
    documentclass: article
    geometry: margin=2.5cm
    fontsize: 11pt
    number-sections: true
jupyter: python3
references:
  - id: golubWelsch
    type: article-journal
    author:
      - family: Golub
        given: Gene H.
      - family: Welsch
        given: John H.
    title: Calculation of Gauss Quadrature Rules
    container-title: Mathematics of Computation
    volume: 23
    issue: 106
    page: 221-230
    issued:
      year: 1969
  - id: cenanovic
    type: webpage
    author:
      - family: Cenanovic
        given: Mirza
      - family: Jansson
        given: Johan
      - family: Jonsson
        given: Carl-Johan
    title: Numerical Integration
    URL: https://python.ju.se/AppliedMaths/Numerical_Integration.html
    issued:
      year: 2024
---
````

Key options explained:

- `title`, `author`, `date`: Document metadata (use `today` for automatic date)
- `format: pdf`: Output format specification
- `documentclass: article`: LaTeX document class (article, report, book)
- `geometry: margin=2.5cm`: Page margins
- `fontsize: 11pt`: Base font size (10pt, 11pt, 12pt)
- `number-sections: true`: Automatically number headings
- `jupyter: python3`: Specify the Jupyter kernel
- `references`: Inline bibliography entries (no external .bib file needed!)

For longer documents, you may want to use an external `.bib` file for references instead of inline entries. See more information in the [Quarto documentation](https://quarto.org/docs/authoring/citations.html).

### Complete Example Report Source


Here is the complete source code for the example report notebook ([example_report.ipynb](example_report.ipynb)):

**Cell 1 (Raw):** YAML Front Matter with inline bibliography
````yaml
---
title: "Numerical Integration: Trapezoidal Rule vs Gaussian Quadrature"
author: "Student Name"
date: today
format:
  pdf:
    documentclass: article
    geometry: margin=2.5cm
    fontsize: 11pt
    number-sections: true
jupyter: python3
references:
  - id: golubWelsch
    type: article-journal
    author:
      - family: Golub
        given: Gene H.
      - family: Welsch
        given: John H.
    title: Calculation of Gauss Quadrature Rules
    container-title: Mathematics of Computation
    volume: 23
    issue: 106
    page: 221-230
    issued:
      year: 1969
  - id: cenanovic
    type: webpage
    author:
      - family: Cenanovic
        given: Mirza
      - family: Jansson
        given: Johan
      - family: Jonsson
        given: Carl-Johan
    title: Numerical Integration
    URL: https://python.ju.se/AppliedMaths/Numerical_Integration.html
    issued:
      year: 2026
---
````



**Cell 2 (Markdown):** Introduction
````markdown
## Introduction

We seek to evaluate the integral

$$
I = \int_{0}^{2} \frac{1}{\sqrt{1 + \sin^2(x)}} \, dx
$$ {#eq-integral}

This integral does not have an elementary closed-form solution. It is an *elliptic integral*[^1]. Therefore, we must use numerical methods. We compare the **Trapezoidal rule** and **Gaussian quadrature** in terms of accuracy and computational efficiency.

[^1]: <https://en.wikipedia.org/wiki/Elliptic_integral>
````



**Cell 3 (Code):** Setup and reference value (hidden with `echo: false` and `output: false`)
````python
#| label: setup
#| echo: false
#| output: false
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate
import time

def f(x):
    return 1 / np.sqrt(1 + np.sin(x)**2)

I_ref, _ = integrate.quad(f, 0, 2, limit=100)
````


**Cell 4 (Markdown):** Method descriptions with equations and citation
````markdown
## Numerical Methods

### Trapezoidal Rule

The composite trapezoidal rule approximates the integral as:

$$
I \approx \frac{h}{2}\left[f(a) + 2\sum_{i=1}^{n-1}f(x_i) + f(b)\right]
$$ {#eq-trapz}

where $h = (b-a)/n$ is the step size.

### Gaussian Quadrature

Gaussian quadrature uses optimally chosen nodes and weights:

$$
I \approx \sum_{i=1}^{n} w_i f(x_i)
$$ {#eq-gauss}

For smooth functions, Gaussian quadrature achieves much higher accuracy with 
fewer function evaluations, see e.g., @golubWelsch and @cenanovic.
````


**Cell 5 (Code):** Convergence plot (log-log scale, hidden code)
````python
#| label: fig-convergence
#| fig-cap: "Convergence comparison: error vs number of points for both methods."
#| echo: false

def trapezoidal(f, a, b, n): 
    x = np.linspace(a, b, n + 1)
    y = f(x)
    h = (b - a) / n
    return h * (0.5 * y[0] + np.sum(y[1:-1]) + 0.5 * y[-1])

def gauss_legendre(f, a, b, n):
    nodes, weights = np.polynomial.legendre.leggauss(n)
    x = 0.5 * (b - a) * nodes + 0.5 * (b + a)
    return 0.5 * (b - a) * np.sum(weights * f(x))

# Convergence study - cut at 25000 iterations
n_trap = np.unique(np.logspace(1, np.log10(25000), 50).astype(int))
n_gauss = np.arange(2, 40)

err_trap = [abs(trapezoidal(f, 0, 2, n) - I_ref) for n in n_trap]
err_gauss = [abs(gauss_legendre(f, 0, 2, n) - I_ref) for n in n_gauss]

plt.figure(figsize=(8, 5))
plt.loglog(n_trap, err_trap, 'b-o', label='Trapezoidal', markersize=4)
plt.loglog(n_gauss, err_gauss, 'r-s', label='Gauss-Legendre', markersize=4)
plt.axhline(1e-10, color='k', linestyle='--', label='Target: $10^{-10}$')
plt.xlabel('Number of points')
plt.ylabel('Absolute error')
plt.legend()
plt.grid(True, alpha=0.3, which='both')
plt.tight_layout()
plt.show()
````



**Cell 6 (Code):** Results table (hidden code, table caption)
````python
#| label: tbl-results
#| tbl-cap: "Points required and computation time to achieve error < $10^{-10}$."
#| echo: false

from IPython.display import Markdown

target = 1e-10
for n in range(10, 100000):
    if abs(trapezoidal(f, 0, 2, n) - I_ref) < target:
        n_trap_min = n
        break

for n in range(2, 100):
    if abs(gauss_legendre(f, 0, 2, n) - I_ref) < target:
        n_gauss_min = n
        break

t0 = time.perf_counter()
for _ in range(1000):
    trapezoidal(f, 0, 2, n_trap_min)
t_trap = (time.perf_counter() - t0) / 1000 * 1e6

t0 = time.perf_counter()
for _ in range(1000):
    gauss_legendre(f, 0, 2, n_gauss_min)
t_gauss = (time.perf_counter() - t0) / 1000 * 1e6

table = f"""
| Method | Points Required | Time (μs) |
|--------|-----------------|----------:|
| Trapezoidal | {n_trap_min} | {t_trap:.1f} |
| Gauss-Legendre | {n_gauss_min} | {t_gauss:.1f} |
"""
Markdown(table)
````



**Cell 7 (Markdown):** Conclusion with cross-references
````markdown
## Conclusion

Gaussian quadrature dramatically outperforms the trapezoidal rule for this smooth
integrand. To achieve an error below $10^{-10}$, trapezoidal rule requires 
thousands of points whereas Gaussian rule requires only 12 points. See 
@fig-convergence for the convergence comparison and @tbl-results for 
the numerical results.

This demonstrates the power of Gaussian quadrature for smooth functions, 
where it achieves *exponential* convergence compared to the *algebraic* 
($O(h^2)$) convergence of the trapezoidal rule.
````
